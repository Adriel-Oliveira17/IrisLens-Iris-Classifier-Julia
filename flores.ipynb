```julia
using MLJ, DataFrames, Random

# Carregar os dados
X, y = @load_iris

df = DataFrame(X)
df.especie = y

first(df, 10)
```

```julia
# Split 80/20, estratificado por classe e embaralhado
train, test = partition(eachindex(y), 0.8, shuffle = true, stratify = y, rng = 42)

# Criando o modelo
DecisionTreeClassifier = @load DecisionTreeClassifier pkg = DecisionTree verbosity = 0
model = DecisionTreeClassifier(max_depth = 3)
mach = machine(model, X, y)

# Treinar o modelo
fit!(mach, rows = train, verbosity = 0)
```

```julia
y_pred = predict_mode(mach, rows = test)
y_true = y[test]

function relatorio_metricas(y_pred, y_true)
    classes = unique(y_true)
    acuracia = sum(y_pred .== y_true) / length(y_true)


    println("Acurácia: ", round(acuracia * 100, digits = 2), "%\n")
    println("Precisão e Recall do modelo")



    for c in classes
        tp = sum((y_pred .== c) .& (y_true .== c))
        fp = sum((y_pred .== c) .& (y_true .!= c))
        fn = sum((y_pred .!= c) .& (y_true .== c))
        precisao = tp + fp == 0 ? 0.0 : tp / (tp + fp)
        recall   = tp + fn == 0 ? 0.0 : tp / (tp + fn)

        

        println(c)
        println("Precisão: ", round(precisao * 100, digits = 2), "%")
        println("Recall:    ", round(recall * 100, digits = 2), "%\n")
    end
end

relatorio_metricas(y_pred, y_true)
```
